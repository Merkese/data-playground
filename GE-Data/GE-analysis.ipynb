```python
import kagglehub as kh
from kagglehub import KaggleDatasetAdapter as kda
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# source path and dataframe
path = "sadiqguru/s-and-p-500-stock-data-along-with-financials-and-news"

filepath = "price_data/GE.csv"

raw_GE_data = kh.dataset_load(
    kda.PANDAS, path, filepath
    )
```

```python
raw_GE_data.head()
```

```python
# checking for none adjusted close column
ge = raw_GE_data.sort_values('Date').copy() 

ge['Date'] = pd.to_datetime(ge['Date'])

```

```python
# Checks percentage change
ge['ret'] = ge['Close'].pct_change()

# checks stats on ret for any inconsistencies
print(ge['ret'].describe())

# checks for largest ret (+ or -) of split smoothness
print(ge.loc[ge['ret'].abs().nlargest(5).index, ['Date', 'Close', 'ret', 'Volume']]) 

# We now know that the close column is split-adjusted, so the pricing data is accurate as far as the 1-for-8 reverse split

# Note: The unadjusted version would have shown a 700% increase
```
